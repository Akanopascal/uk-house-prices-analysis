```python
# A list of names, in the same order as columns 0–15
column_names = [
    "transaction_id", "price", "date", "postcode",
    "property_type", "new_build", "tenure",
    "paon", "saon", "street", "locality",
    "town", "district", "county",
    "category", "record_status"
]

# Replace the numbered labels (0, 1, 2...) with our names
df.columns = column_names

# Check it worked
df.head()
```

```python
# Turn the text dates into real dates
df["date"] = pd.to_datetime(df["date"])

# Check the new type
df["date"].dtype
```

```python
# Earliest and latest sale dates
print("First sale:", df["date"].min())
print("Last sale:", df["date"].max())
```

```python
# The 10 cheapest sales
df.nsmallest(10, "price")
```

```python
# The 10 cheapest sales
df.nlargest(10, "price")
```

```python
# Keep only typical home sales
homes = df[
    (df["category"] == "A") &          # standard sales only
    (df["property_type"] != "O") &     # houses and flats only, no "other"
    (df["price"] >= 10000)             # remove nominal sales like £1
]

# How much did we keep?
print("Before:", len(df))
print("After:", len(homes))
print("Removed:", len(df) - len(homes))
```

```python
homes["price"].describe()
```

## Cleaning decisions

- Kept only **category A** (standard sales); removed repossessions, buy-to-lets and company purchases
- Removed **property type O** (hotels, offices, parking spaces) to focus on homes
- Removed sales **under £10,000** (nominal transfers like £1)
- Result: 822,646 of 992,278 sales kept (83%)
- Note: category A alone did almost all the filtering; the other two rules removed only 8 extra rows
- Will use the **median** for comparisons, as prices are skewed by a few very expensive homes

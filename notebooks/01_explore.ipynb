## Data dictionary (my investigation, checked against HM Land Registry guide)

| Column | Name | Meaning | My guess was... |
|---|---|---|---|
| 0 | transaction_id | Unique ID for each sale | ✅ Correct |
| 1 | price | Sale price (£) | ✅ Correct |
| 2 | date | Date the sale completed | ✅ Correct |
| 3 | postcode | Property postcode | ✅ Correct |
| 4 | property_type | D = detached, S = semi, T = terraced, F = flat, O = other | ✅ Correct (O was a hint) |
| 5 | new_build | Y = new build, N = existing | ✅ Correct |
| 6 | tenure | F = freehold, L = leasehold | ✅ Correct |
| 7 | paon | House number or building name | ⚠️ Half (thought: court names) |
| 8 | saon | Flat or unit number | ✅ Correct |
| 9 | street | Street name | ✅ Correct |
| 10 | locality | Neighbourhood within a town | ❌ (thought: city) |
| 11 | town | Town or city | ✅ Correct |
| 12 | district | Local council area | ❌ (thought: county) |
| 13 | county | County | ✅ Correct |
| 14 | category | A = standard sale, B = additional (repossessions, buy-to-let, company sales) | ✅ Close |
| 15 | record_status | A = addition, C = change, D = deletion (monthly files only) | Not guessed |
